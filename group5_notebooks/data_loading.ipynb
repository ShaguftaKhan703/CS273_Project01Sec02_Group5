## 1️⃣ **Create Classes for Data Samples** ##

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
# Load the preprocessed data
emo_sounds_df = pd.read_csv("/content/drive/MyDrive/CS273_Project01Sec02_Group5/group5_preprocessed_data/clean_emo_sounds.csv", index_col=0)
iadsed_df = pd.read_csv("/content/drive/MyDrive/CS273_Project01Sec02_Group5/group5_preprocessed_data/IADSED_preprocessed.csv")
```

```python
# Create classes: map the target values and label each sample based on the polarity of arousal and valence
scaler = StandardScaler()

emo_sounds_df[["arousal", "valence"]] = scaler.fit_transform(emo_sounds_df[["arousal", "valence"]])
iadsed_df[["arousal", "valence"]] = scaler.fit_transform(iadsed_df[["arousal", "valence"]])

def create_classes(row):
  if row["arousal"] > 0 and row["valence"] > 0:
    val = "class 1"

  elif row["arousal"] > 0 and row["valence"] < 0:
    val = "class 2"

  elif row["arousal"] < 0 and row["valence"] < 0:
    val = "class 3"

  else:
    val = "class 4"


  return val


emo_sounds_df["class"] = emo_sounds_df.apply(create_classes, axis=1)
iadsed_df["class"] = iadsed_df.apply(create_classes, axis=1)
```

```python
# Save the final datasets
emo_sounds_df.to_csv("/content/drive/MyDrive/CS273_Project01Sec02_Group5/group5_preprocessed_data/final_emosounds.csv")
iadsed_df.to_csv("/content/drive/MyDrive/CS273_Project01Sec02_Group5/group5_preprocessed_data/final_iadsed.csv")
```
